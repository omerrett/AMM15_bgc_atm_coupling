```python
import xarray as xr
import datetime
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FuncFormatter
```

```python
AOWB_onset = xr.open_dataset(
'/gws/pw/j25/rep/omerrett/AOWB_RALP3P3_fixes/Analysis/bloom_onset/10m_depth/bloom_onset_5day_15growth_chloro_top_10m.nc'
)
OWB_onset = xr.open_dataset(
'/gws/pw/j25/rep/omerrett/OWB/Analysis/Bloom_Onset/10m_depth/bloom_onset_5day_15growth_chloro_top_10m.nc'
)

diff_onset = AOWB_onset - OWB_onset
```

```python

# Use the same plotting style throughout all plots
plt.style.use('seaborn-v0_8-whitegrid')

# Extent and projection for AMM15
TIGHT_EXTENT = [-18.5, 8, 44, 63] 
NORTH_SEA_PROJECTION = ccrs.LambertConformal(
    central_longitude=-5,
    central_latitude=50.0,
    standard_parallels=(50.0, 60.0)
)

# This is used for colourbar ticks
def day_to_date_2018(x, pos):
    """Converts Day of Year integer to 'Month Day' string for 2018."""
    try:
        date = datetime.datetime(2018, 1, 1) + datetime.timedelta(days=int(x) - 1)
        return date.strftime('%b %d')
    except:
        return ""

# Align coordinates
diff_onset = diff_onset.assign_coords({"nav_lon": AOWB_onset.nav_lon, "nav_lat": AOWB_onset.nav_lat})

# Define the three fields and their titles to match template list structure
PLOTS = [
    {'data': OWB_onset['chloro_top_bloom_onset_index'], 'title': 'OWB'},
    {'data': AOWB_onset['chloro_top_bloom_onset_index'], 'title': 'AOWB'},
    {'data': diff_onset['chloro_top_bloom_onset_index'], 'title': 'AOWB - OWB'}
]

fig, axes = plt.subplots(
    ncols=3,
    figsize=(24, 8),
    subplot_kw={'projection': NORTH_SEA_PROJECTION},
    dpi=600
)
axes = np.atleast_1d(axes)

vmins = [89, 89, -28] 
vmaxs = [190, 190, 28]
cmaps = ['YlGnBu_r', 'YlGnBu_r', 'Spectral_r']

for i, plot_config in enumerate(PLOTS):
    ax = axes[i]
    data = plot_config['data']
    cmap = cmaps[i]
    vmin = vmins[i]
    vmax = vmaxs[i]

    ax.add_feature(cfeature.LAND, facecolor='#E6E6E6', zorder=1) 
    
    # Plot the data using the exact structural formatting
    p = data.plot(
        ax=ax,
        x='nav_lon',
        y='nav_lat',
        transform=ccrs.PlateCarree(),
        cmap=cmap, 
        vmin=vmin, 
        vmax=vmax,
        add_colorbar=False,
        zorder=2
    )

    # All of this is for colourbar formatting
    cbar = plt.colorbar(
        p,
        ax=ax,
        shrink=0.8,
        aspect=30,
        pad=0.02
    )
    
    cbar.outline.set_edgecolor('black')
    cbar.outline.set_linewidth(1)
    cbar.ax.tick_params(labelsize=15)

    if i == 0:
        cbar.ax.set_visible(False)
        cbar.outline.set_visible(False)
        
    elif i == 1:
        cbar.ax.yaxis.set_major_formatter(FuncFormatter(day_to_date_2018))
        ticks = [90, 110, 130, 150, 171, 191]
        cbar.set_ticks(ticks)
        cbar.set_label('Date of Bloom Onset', fontsize=20)
        
    elif i == 2:
        ticks = [-21, -14, -7, 0, 7, 14, 21]
        cbar.set_ticks(ticks)
        cbar.set_ticklabels([str(t) for t in ticks])
        cbar.set_label('Bloom Onset Delay (Days)', fontsize=20)

    ax.set_extent(TIGHT_EXTENT, crs=ccrs.PlateCarree()) 
    ax.coastlines(resolution='50m', linewidth=1.2, color='grey', zorder=3)
    ax.gridlines(draw_labels=False, color='grey', linestyle='--') 

    if i == 2:
        ax.text(
        0.985, 0.02, 
        plot_config['title'],
        transform=ax.transAxes, 
        fontsize=20, 
        fontweight='bold',
        color='black',
        ha='right',
        zorder=5
    )

    else:
        ax.text(
        0.985, 0.02, 
        plot_config['title'],
        transform=ax.transAxes, 
        fontsize=24, 
        fontweight='bold',
        color='black',
        ha='right',
        zorder=5
    )
    

    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_title('')

    for spine in ax.spines.values():
        spine.set_edgecolor('black')
        spine.set_linewidth(1)

label_style = dict(fontsize=24, fontweight='bold', va='top', ha='left', zorder=6)
axes[0].text(0.01, 0.99, 'a)', transform=axes[0].transAxes, **label_style)
axes[1].text(0.01, 0.99, 'b)', transform=axes[1].transAxes, **label_style)
axes[2].text(0.01, 0.99, 'c)', transform=axes[2].transAxes, **label_style)

plt.tight_layout()
plt.savefig('Bloom_Onset_Comparison_Paper_Style.png', dpi=600, bbox_inches='tight')
plt.show()
```
